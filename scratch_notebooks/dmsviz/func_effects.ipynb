```python
import pandas as pd
```

```python
func_effects_23 = pd.read_csv('../../results/func_effects/averages/293_2-3_entry_func_effects.csv')
func_effects_23
```

```python
func_effects_26 = pd.read_csv('../../results/func_effects/averages/293_2-6_entry_func_effects.csv')
func_effects_26
```

```python
map = pd.read_csv('site_numbering_map.csv')
map
```

```python
func_effects_ref_23 = pd.merge(
    func_effects_23,
    map,
    left_on=['site'],
    right_on=['reference_site']
)[['sequential_site', 'wildtype', 'mutant', 'effect', 'chain']].rename(
    columns={'sequential_site' : 'site'}
)

func_effects_ref_23.to_csv('2-3_entry_func_effects_sequential.csv')
```

```python
func_effects_ref_26 = pd.merge(
    func_effects_26,
    map,
    left_on=['site'],
    right_on=['reference_site']
)[['sequential_site', 'wildtype', 'mutant', 'effect', 'chain']].rename(
    columns={'sequential_site' : 'site'}
)

func_effects_ref_26.to_csv('2-6_entry_func_effects_sequential.csv')
```
