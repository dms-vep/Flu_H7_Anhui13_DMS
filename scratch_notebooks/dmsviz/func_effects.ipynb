```python
import pandas as pd
```

```python
func_effects = pd.read_csv('293T_entry_func_effects.csv')
func_effects
```

```python
map = pd.read_csv('site_numbering_map.csv')
map
```

```python
func_effects_ref = pd.merge(
    func_effects,
    map,
    left_on=['site'],
    right_on=['reference_site']
)[['condition', 'sequential_site', 'wildtype', 'mutant', 'effect', 'chain']].rename(
    columns={'sequential_site' : 'site'}
)

func_effects_ref.to_csv('293T_entry_func_effects_sequential.csv')
```
